# RL Entry/Hedge Trainer

This notebook generates a sample dataset, loads it, and trains the Q-learning agent.


```python
import csv
from pathlib import Path

data_dir = Path('data')
data_dir.mkdir(exist_ok=True)
csv_path = data_dir / 'market_snapshots.csv'

rows = []
base_ts = 1700000000
for i in range(200):
    ts = base_ts + i * 60
    polymarket_bid = 0.45 + (i % 10) * 0.002
    polymarket_ask = polymarket_bid + 0.02
    chainlink_price = 25000 + i * 2
    binance_price = 25000 + i * 2 + (-5 + (i % 5) * 2)
    binance_cvd = 1000 + (i * 10) - (i % 7) * 15
    rows.append([
        ts,
        round(polymarket_bid, 4),
        round(polymarket_ask, 4),
        chainlink_price,
        binance_price,
        binance_cvd,
    ])

with open(csv_path, 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['timestamp', 'polymarket_bid', 'polymarket_ask', 'chainlink_price', 'binance_price', 'binance_cvd'])
    writer.writerows(rows)

csv_path
```

```python
from ml import rl_entry_hedge

snapshots = rl_entry_hedge.load_snapshots_from_csv(str(csv_path))
agent = rl_entry_hedge.train_agent(snapshots, episodes=3, log_every_steps=50)
len(agent.q_table)
```
